```python
from openai import OpenAI
import os
from dotenv import load_dotenv

load_dotenv()

client = OpenAI(api_key=os.getenv("OpenAI_API_KEY"))
```

```python
sample_text = """
The product quality was excellent , but the delivery was kinda delayed.
Customer support respond quickly and resolve the issue

"""

def run_task(prompt,text):
    '''
    #asks the model to generate a reply
    response = client.chat.completions.create(
        #using ml
        model="gpt-3.5-turbo",
        messages=[
            {"role":"user" , "content":f"{prompt}\n\nText:\n{text}"}
        ],
        temperature=0.5
    )
    return response.choices[0].message.content
    '''
    return "Response"

def run_exp(prompt_file):
    with open(prompt_file) as f:
        prompts = f.readlines()
    result = []
    for prompt in prompts:
        output = run_task(prompt.strip(),sample_text)
        result.append((prompt.strip(),output))
    return result

basic_result = run_exp("../prompts/basic_prompt.txt")
structured_reult = run_exp("../prompts/structured_prompts.txt")

for p , o in basic_result:
    print(p)
    print(o)
    print("_"*12)

for p , o in structured_reult:
    print(p)
    print(o)
    print("_"*12)
```
