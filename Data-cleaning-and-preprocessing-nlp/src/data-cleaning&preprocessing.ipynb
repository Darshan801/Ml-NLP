```python
import pandas as pd

df= pd.read_csv('../data/customer_data.csv')
df.head(10)
```

data understanding
info() = column names, data types, missing values

describe() = summary statistics

```python
print(df.shape)
print(df.info())
print(df.describe(include='all'))
```

handaling missing values

```python
import numpy as np
df.replace(r'^\s*$', np.nan , regex=True)

df['Age']  = pd.to_numeric(df['Age'],errors='coerce')
df['Age']  = df['Age'].fillna(df['Age'].mean())
df.loc[(df['Age'] < 0) | (df['Age'] > 100), 'Age'] = df['Age'].mean()


df['Name']  = df['Name'].fillna('Unknown')

df['Purchase_Amount'] = pd.to_numeric(df['Purchase_Amount'], errors='coerce')
df['Purchase_Amount'] = df['Purchase_Amount'].fillna(df['Purchase_Amount'].mean())

df['Last_Login'] = df['Last_Login'].fillna(df['Last_Login'].mode()[0])

df = df.drop_duplicates()

print(df.head())
```

Gender Normalization

```python
df['Gender'] = df['Gender'].replace({
    'M':'Male',
    'm':'Male',
    'F':"Female",
    'f':"Female",
})
print(df.head())
```

email validation

```python
import re
import pandas as pd

def validate_email(email):
    if pd.isna(email):
        return np.nan
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    return email if re.match(pattern,email) else np.nan
df['Email'] = df['Email'].apply(validate_email)
print(df.head())
```

data processing

```python
df['Signup_Date'] = pd.to_datetime(df['Signup_Date'],errors='coerce')
df['Last_Login'] = pd.to_datetime(df['Last_Login'],errors='coerce')

df['Signup_Date'] = df['Signup_Date'].fillna(df['Signup_Date'].mode()[0])
df['Last_Login'] = df['Last_Login'].fillna(df['Last_Login'].mode()[0])
print(df.head())
```

feature extraction

outlier detection

```python
#calculate quartile rannge
Q1 = df['Purchase_Amount'].quantile(0.25)
Q3 = df['Purchase_Amount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df = df[
    (df['Purchase_Amount'] >= lower_bound) &
    (df['Purchase_Amount'] <= upper_bound)
]
```

text standardizaton

```python
df['Country'] = df['Country'].str.strip().str.title()
df['Status'] = df['Status'].str.strip().str.title()
print(df.head())
```

Dropping irrelevent columns

```python
df.drop(['Name', 'Email', 'Signup_Date'], axis=1, inplace=True)
```
